# V3 SAM + AdamW: Dual EMA-best / Raw-best, HC3 → GPT-3.5 0%

Notebook mới dựa trên bản 07 Dual. **Bật GPU + Internet rồi Run All.**

- Mặc định seed **2020**, train batch = micro-batch = **1**, validation batch **64**, 10 epoch, `struct_mode=low`, `select=lossaug`, threshold test **0.5**.
- SAM `rho=0.05` bọc AdamW; không đổi model, encoder, graph mapping hoặc cách chọn hai checkpoint độc lập.
- Mỗi optimizer update: backward clean/swap → nhiễu trọng số → replay RNG → backward clean/swap lần 2 → khôi phục trọng số → AdamW → scheduler/EMA **một lần**.
- Dropout, edge-drop và swap được replay cùng RNG giữa hai lượt. First-pass gradient không clip trước khi tính hướng nhiễu; second-pass gradient clip 1.0 như baseline.
- AMP **tắt** (`--amp 0`) để không có tương tác GradScaler/SAM chưa được kiểm chứng. `SAM_RHO=0` là đối chứng AdamW một lượt.
- Train loss là loss tại trọng số gốc; `sam_perturbed_loss` là loss tại trọng số nhiễu. Checkpoint không lưu trọng số đang bị perturb.
- Giữ deterministic. Notebook nhúng helper từ nguồn local hiện tại, không yêu cầu upload ZIP baseline hay sửa model trên GitHub.
- Mỗi run có thư mục riêng, không ghi đè baseline. ZIP cuối chứa checkpoint, metadata, mã nguồn, config và kết quả test. Không dùng test để chọn checkpoint.

SAM tăng chi phí tính gradient khoảng hai lần và cần thêm bộ nhớ lưu bản sao trọng số. Đây là thử nghiệm cải thiện tổng quát hóa, không đảm bảo tăng accuracy.

```python
# Self-contained SAM trainer; original model architecture is unchanged.
DUAL_TRAINER=REPO/'train_v3_sam_dual_best.py'
DUAL_TRAINER.write_text('"""Train the Quang V3 Core from legacy-domain graphs.\n\nNo test files are read during training or checkpoint selection.\n"""\n\nimport argparse\nimport os\nimport copy\nimport gc\nimport hashlib\nimport json\nimport math\nfrom pathlib import Path\nimport time\n\nfrom evaluate_v3_quang import (REPO, TRAINING_DOMAINS, environment, positive_int,\n                               sha256, write_json)\n\n\ndef configure_determinism(enabled=True):\n    """Configure this process before any CUDA work, not only the notebook parent.\n\n    Seed selection remains in set_seed(). Unsupported deterministic operations\n    raise an error instead of silently running a nondeterministic alternative.\n    """\n    import torch\n\n    if enabled:\n        workspace = os.environ.get("CUBLAS_WORKSPACE_CONFIG")\n        if workspace not in (":4096:8", ":16:8"):\n            if torch.cuda.is_initialized():\n                raise RuntimeError("Enable determinism before initializing CUDA; restart the Python process")\n            os.environ["CUBLAS_WORKSPACE_CONFIG"] = ":4096:8"\n    torch.use_deterministic_algorithms(enabled, warn_only=False)\n    torch.backends.cudnn.benchmark = False\n    torch.backends.cudnn.deterministic = enabled\n    return {\n        "enabled": torch.are_deterministic_algorithms_enabled(),\n        "warn_only": torch.is_deterministic_algorithms_warn_only_enabled(),\n        "cublas_workspace_config": os.environ.get("CUBLAS_WORKSPACE_CONFIG"),\n        "cudnn_benchmark": torch.backends.cudnn.benchmark,\n        "cudnn_deterministic": torch.backends.cudnn.deterministic,\n    }\n\n\n\ndef capture_sam_rng(device):\n    import torch\n    return (torch.get_rng_state().clone(),\n            torch.cuda.get_rng_state(device).clone() if device.type == "cuda" else None)\n\n\ndef restore_sam_rng(state, device):\n    import torch\n    torch.set_rng_state(state[0])\n    if state[1] is not None:\n        torch.cuda.set_rng_state(state[1], device)\n\n\ndef backward_objective(model, dataset, indices, device, gm, gs, micro_size, aug):\n    """Same clean/swap BCE and accumulation weights as the non-SAM trainer."""\n    import torch\n    import torch.nn.functional as F\n    total_loss = 0.0\n    for micro in indices.split(micro_size):\n        batch = dataset.batch(micro, device, gm, gs)\n        weight = batch["B"] / indices.numel()\n        branch_weight = 0.5 if aug != "none" else 1.0\n        logits = model(batch)\n        loss = F.binary_cross_entropy_with_logits(logits.float(), batch["y"])\n        if not torch.isfinite(loss):\n            raise FloatingPointError("Non-finite clean loss in SAM")\n        (weight * branch_weight * loss).backward()\n        total_loss += weight * branch_weight * float(loss.detach())\n        del logits, loss\n        if aug != "none":\n            logits = model(batch, aug_mode=aug)\n            loss = F.binary_cross_entropy_with_logits(logits.float(), batch["y"])\n            if not torch.isfinite(loss):\n                raise FloatingPointError("Non-finite augmented loss in SAM")\n            (weight * 0.5 * loss).backward()\n            total_loss += weight * 0.5 * float(loss.detach())\n            del logits, loss\n        del batch\n    return total_loss\n\n\ndef sam_adamw_update(model, optimizer, objective, device, rho):\n    """Two passes, one AdamW step; restore exact weights even if pass 2 fails.\n\n    Perturb with the un-clipped first gradient. Replay CPU/CUDA RNG so dropout,\n    edge-drop and syntax swap are identical across the two objective passes.\n    The caller updates its LR schedule and EMA only after this function returns.\n    rho=0 is an explicit one-pass AdamW control for checking the implementation.\n    """\n    import torch\n    if not math.isfinite(rho) or rho < 0:\n        raise ValueError("SAM rho must be finite and non-negative")\n    optimizer.zero_grad(set_to_none=True)\n    initial_rng = capture_sam_rng(device)\n    clean_objective = objective()\n    params = [p for p in model.parameters() if p.grad is not None]\n    if not params:\n        raise RuntimeError("SAM objective produced no parameter gradients")\n    grad_norm = torch.linalg.vector_norm(torch.stack([p.grad.detach().norm(2) for p in params]), 2)\n    if not torch.isfinite(grad_norm):\n        raise FloatingPointError("Non-finite first-pass SAM gradient")\n    perturbed_objective = clean_objective\n    if rho > 0:\n        originals = [p.detach().clone() for p in params]\n        try:\n            with torch.no_grad():\n                scale = rho / (grad_norm + 1e-12)\n                for p in params:\n                    p.add_(p.grad, alpha=float(scale))\n            optimizer.zero_grad(set_to_none=True)\n            restore_sam_rng(initial_rng, device)\n            perturbed_objective = objective()\n        finally:\n            # Copy, not subtraction: avoid accumulating round-off restoration error.\n            with torch.no_grad():\n                for p, original in zip(params, originals):\n                    p.copy_(original)\n            del originals\n    torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0, error_if_nonfinite=True)\n    optimizer.step()\n    return clean_objective, perturbed_objective, float(grad_norm)\n\n\ndef build_parser():\n    parser = argparse.ArgumentParser(description=__doc__)\n    parser.add_argument("--variants", nargs="+", choices=("baseline",), default=["baseline"],\n                        help="Retained for CLI compatibility; V3 Core has one baseline variant.")\n    parser.add_argument("--seeds", type=int, nargs="+", default=[2025, 2026])\n    parser.add_argument("--graph-dir", type=Path, default=REPO / "graph_data")\n    parser.add_argument("--output-dir", type=Path, default=REPO / "result/v3_baseline/model")\n    parser.add_argument("--train-file", default="hc3_train")\n    parser.add_argument("--val-file", default="hc3_val")\n    parser.add_argument("--struct-mode", choices=("low", "high"), default="low")\n    parser.add_argument("--num-rel", type=int, default=0, help="0: infer from training graph only")\n    parser.add_argument("--hidden", type=positive_int, default=384)\n    parser.add_argument("--num-bases", type=positive_int, default=12)\n    parser.add_argument("--layers", type=positive_int, default=2)\n    parser.add_argument("--dropout", type=float, default=0.3)\n    parser.add_argument("--in-dropout", type=float, default=0.1)\n    parser.add_argument("--edge-drop", type=float, default=0.1)\n    parser.add_argument("--struct", type=int, choices=(0, 1), default=1)\n    parser.add_argument("--gfeat", type=int, choices=(0, 1), default=0)\n    parser.add_argument("--readout", choices=("pool3",), default="pool3")\n    parser.add_argument("--aug", choices=("none", "swap", "mask", "both"), default="swap")\n    parser.add_argument("--aug-max", type=float, default=0.35)\n    parser.add_argument("--ema", type=int, choices=(0, 1), default=1)\n    parser.add_argument("--ema-decay", type=float, default=0.998)\n    parser.add_argument("--sam-rho", type=float, default=0.05)\n    parser.add_argument("--lr", type=float, default=5e-4)\n    parser.add_argument("--wd", type=float, default=0.01)\n    parser.add_argument("--epochs", type=positive_int, default=10)\n    parser.add_argument("--patience", type=positive_int, default=10)\n    parser.add_argument("--batch-size", type=positive_int, default=32,\n                        help="Effective graphs per optimizer/EMA/scheduler step")\n    parser.add_argument("--micro-batch-size", type=positive_int, default=8,\n                        help="Graph donor pool size; use the same value in both variants")\n    parser.add_argument("--val-batch-size", "--eval-batch-size", dest="val_batch_size",\n                        type=positive_int, default=64,\n                        help="Also controls the donor pool of the fixed validation swap proxy")\n    parser.add_argument("--select", choices=("lossaug", "loss", "acc", "auc"), default="lossaug")\n    parser.add_argument("--amp", type=int, choices=(0, 1), default=0)\n    parser.add_argument("--device", choices=("cpu", "cuda"), default="cuda")\n    parser.add_argument("--deterministic", action=argparse.BooleanOptionalAction, default=True,\n                        help="Use reproducible algorithms (default: on); --no-deterministic restores legacy behavior")\n    parser.add_argument("--skip-completed", action="store_true")\n    parser.add_argument("--overwrite", action="store_true")\n    return parser\n\n\ndef validate_args(args):\n    if args.amp:\n        raise ValueError("This SAM experiment requires --amp 0")\n    if not math.isfinite(args.sam_rho) or args.sam_rho < 0:\n        raise ValueError("--sam-rho must be finite and non-negative")\n    pairs = {(item["train_file"], item["val_file"]) for item in TRAINING_DOMAINS.values()}\n    if (args.train_file, args.val_file) not in pairs:\n        raise ValueError("Use a supported train/validation pair from TRAINING_DOMAINS")\n    if len(set(args.seeds)) != len(args.seeds) or len(set(args.variants)) != len(args.variants):\n        raise ValueError("Seeds/variants must not contain duplicates")\n    for name in ("dropout", "in_dropout", "edge_drop", "aug_max"):\n        if not 0 <= getattr(args, name) <= 1:\n            raise ValueError(f"{name} must be in [0,1]")\n    for name in ("lr",):\n        if not math.isfinite(getattr(args, name)) or getattr(args, name) <= 0:\n            raise ValueError(f"{name} must be finite and positive")\n    if not math.isfinite(args.wd) or args.wd < 0 or not 0 <= args.ema_decay < 1 or args.num_rel < 0:\n        raise ValueError("Invalid weight decay, EMA decay or relation count")\n    if args.micro_batch_size > args.batch_size:\n        raise ValueError("micro-batch-size must not exceed batch-size")\n\n\ndef effective_readout(args, variant):\n    return args.readout\n\n\ndef build_model(args, variant, R):\n    from model.QuangV3 import Net\n\n    kwargs = dict(d=args.hidden, nb=args.num_bases, L=args.layers, p=args.dropout,\n                  p_in=args.in_dropout, edge_drop=args.edge_drop,\n                  use_struct=bool(args.struct), use_gfeat=bool(args.gfeat),\n                  readout=effective_readout(args, variant), aug=args.aug,\n                  aug_max=args.aug_max, mode=args.struct_mode)\n    return Net(R, **kwargs)\n\n\ndef load_graph(path, source, R, mode):\n    """Validate original cache schema and exact source label order, then keep on CPU."""\n    import numpy as np\n    import torch\n    from model import QuangV3 as v3\n\n    original_load = v3._load_pickle\n\n    def checked_load(filename):\n        raw = original_load(filename)\n        if not isinstance(raw, dict):\n            raise ValueError("Use dictionary caches from building_graph.py")\n        if any(key in raw for key in ("encoder_metadata", "schema_version", "all_cls_embeddings",\n                                      "all_quality", "quality_features")):\n            raise ValueError("Use legacy V1 graphs, not aligned/V2 caches")\n        for x, edge, rel in zip(raw["all_token_embeddings"], raw["all_edge_index"], raw["all_edge_type"]):\n            edge, rel = torch.as_tensor(edge), torch.as_tensor(rel)\n            if x.ndim != 2 or x.shape[0] == 0 or x.shape[1] != 768:\n                raise ValueError("Invalid legacy node embeddings; do not use aligned/V2 graphs")\n            if edge.shape != (2, rel.numel()) or rel.numel() == 0:\n                raise ValueError("Invalid relation/edge shapes")\n            if int(rel.min()) < 0 or int(edge.min()) < 0 or int(edge.max()) >= x.shape[0]:\n                raise ValueError("Invalid relation/edge indices")\n            if R and int(rel.max()) >= R:\n                raise ValueError("Validation relation IDs exceed training R; refusing silent clipping")\n        sizes = [len(raw[k]) for k in ("all_token_embeddings", "all_edge_index", "all_edge_type", "y")]\n        if len(set(sizes)) != 1:\n            raise ValueError("Cache arrays have different numbers of samples")\n        return raw\n\n    v3._load_pickle = checked_load\n    try:\n        ds = v3.GraphSet(str(path), R=R or None, mode=mode, store_dev="cpu")\n    finally:\n        v3._load_pickle = original_load\n    expected = []\n    with source.open(encoding="utf-8") as stream:\n        for line in stream:\n            if line.strip():\n                label = json.loads(line)["label"]\n                expected.append(int(label) if isinstance(label, (int, float)) else int("human" in str(label).lower()))\n    if not np.array_equal(ds.y.numpy(), expected) or set(expected) != {0, 1}:\n        raise ValueError("Cache/source labels or counts differ; rebuild with building_graph.py")\n    return ds\n\n\ndef validation_loss(probabilities, truth):\n    import numpy as np\n\n    p = np.clip(probabilities, 1e-6, 1 - 1e-6)\n    return float(-(truth * np.log(p) + (1 - truth) * np.log(1 - p)).mean())\n\n\ndef atomic_save(payload, path):\n    import torch\n\n    temporary = path.with_name(path.name + ".tmp")\n    torch.save(payload, temporary)\n    temporary.replace(path)\n\n\ndef signature(args, variant, seed, data_hashes):\n    config = vars(args).copy()\n    for key in ("variants", "seeds", "output_dir", "graph_dir", "skip_completed", "overwrite"):\n        config.pop(key)\n    config["effective_readout"] = effective_readout(args, variant)\n    code = {str(p.relative_to(REPO)): sha256(p) for p in (\n        Path(__file__), REPO / "train_v3.py", REPO / "model/QuangV3.py", REPO / "building_graph.py", REPO / "dep_vocab.py",\n        REPO / "dep_vocab.json",\n    )}\n    record = dict(config=config, variant=variant, seed=seed, data=data_hashes,\n                  code_sha256=code, environment=environment())\n    digest = hashlib.sha256(json.dumps(record, sort_keys=True).encode()).hexdigest()\n    return digest, record\n\n\ndef completed_run(path, record_path, run_signature):\n    if not path.is_file() or not record_path.is_file():\n        return False\n    record = json.loads(record_path.read_text(encoding="utf-8"))\n    raw_path = path.with_name(path.name.replace("_ema_seed", "_raw_seed"))\n    hashes = record.get("checkpoint_sha256", {})\n    return (record.get("complete") is True and record.get("run_signature") == run_signature\n            and raw_path.is_file() and isinstance(hashes, dict)\n            and hashes.get("ema") == sha256(path) and hashes.get("raw") == sha256(raw_path))\n\n\ndef train_one(args, variant, seed, train, val, run_signature, provenance):\n    import numpy as np\n    import torch\n    import torch.nn.functional as F\n    from model.QuangV3 import predict, set_seed\n    from evaluate_v3_quang import score_probabilities\n\n    device = torch.device(args.device)\n    set_seed(seed)\n    model = build_model(args, variant, train.R).to(device)\n    teacher = copy.deepcopy(model).eval() if args.ema else None\n    optimizer = torch.optim.AdamW(model.parameters(), lr=args.lr, weight_decay=args.wd)\n    steps_per_epoch = math.ceil(train.G / args.batch_size)\n    total = steps_per_epoch * args.epochs\n    warm = max(1, int(total * 0.05))\n    schedule = torch.optim.lr_scheduler.LambdaLR(\n        optimizer, lambda s: (s + 1) / warm if s < warm else\n        0.5 * (1 + math.cos(math.pi * (s - warm) / max(1, total - warm))) * 0.95 + 0.05,\n    )\n    amp = bool(args.amp) and device.type == "cuda"\n    gm = train.GF.float().mean(0).cpu()\n    gs = train.GF.float().std(0).cpu().clamp_min(1e-6)\n    ema_checkpoint_path = args.output_dir / f"v3_{variant}_ema_seed{seed}.pt"\n    raw_checkpoint_path = args.output_dir / f"v3_{variant}_raw_seed{seed}.pt"\n    record_path = args.output_dir / f"v3_{variant}_dual_seed{seed}.train.json"\n    saved_args = {k: str(v) if isinstance(v, Path) else v for k, v in vars(args).items()}\n    saved_args.update(seeds=[seed], cons=0.0, readout=effective_readout(args, variant))\n    history = []\n    best_ema_score = best_raw_score = float("inf")\n    best_ema_epoch = best_raw_epoch = 0\n    bad = step = 0\n    start = time.time()\n    sampler_generator = torch.Generator(device="cpu").manual_seed(seed)\n    print(f"{variant} seed={seed} | R={train.R} mode={args.struct_mode} | "\n          f"effective batch={args.batch_size}, micro={args.micro_batch_size}", flush=True)\n\n    for epoch in range(1, args.epochs + 1):\n        model.train()\n        permutation = torch.randperm(train.G, generator=sampler_generator)\n        epoch_loss = 0.0\n        epoch_perturbed_loss = 0.0\n        epoch_grad_norm = 0.0\n        epoch_updates = 0\n        for offset in range(0, train.G, args.batch_size):\n            indices = permutation[offset:offset + args.batch_size]\n            def objective():\n                return backward_objective(model, train, indices, device, gm, gs,\n                                          args.micro_batch_size, args.aug)\n\n            base_loss, perturbed_loss, grad_norm = sam_adamw_update(\n                model, optimizer, objective, device, args.sam_rho)\n            epoch_loss += base_loss * indices.numel()\n            epoch_perturbed_loss += perturbed_loss * indices.numel()\n            epoch_grad_norm += grad_norm\n            epoch_updates += 1\n            # Only restored, final AdamW weights are visible to schedule/teacher.\n            schedule.step()\n            step += 1\n            if teacher is not None:\n                decay = min(args.ema_decay, (1 + step) / (10 + step))\n                with torch.no_grad():\n                    for target, current in zip(teacher.parameters(), model.parameters()):\n                        target.mul_(decay).add_(current.detach(), alpha=1 - decay)\n        # Each weight family chooses its own best validation epoch.  The raw\n        # model and EMA teacher therefore receive identical training updates,\n        # but neither is selected using the other model\'s validation scores.\n        ema_model = teacher if teacher is not None else model\n        raw_model = model\n        truth = val.y.numpy().astype(int)\n\n        def validate(candidate):\n            probabilities = predict(candidate, val, device, gm, gs, bs=args.val_batch_size, amp=amp)\n            swapped = predict(candidate, val, device, gm, gs, bs=args.val_batch_size, amp=amp, aug_mode="swap")\n            metrics = score_probabilities(truth, probabilities, 0.5)\n            clean_loss = validation_loss(probabilities, truth)\n            swap_loss = validation_loss(swapped, truth)\n            score = {"lossaug": clean_loss + swap_loss, "loss": clean_loss,\n                     "acc": -metrics["accuracy"], "auc": -metrics["auc"]}[args.select]\n            thresholds = np.linspace(0.05, 0.95, 91)\n            accuracies = np.array([((probabilities >= threshold) == truth).mean() for threshold in thresholds])\n            threshold = float(np.median(thresholds[accuracies >= accuracies.max() - 1e-9]))\n            return probabilities, clean_loss, swap_loss, score, metrics, threshold\n\n        ema_p, ema_clean, ema_swap, ema_score, ema_metrics, ema_threshold = validate(ema_model)\n        raw_p, raw_clean, raw_swap, raw_score, raw_metrics, raw_threshold = validate(raw_model)\n        if (not math.isfinite(epoch_loss) or not math.isfinite(ema_score)\n                or not math.isfinite(raw_score)):\n            raise FloatingPointError("Non-finite loss; no completed checkpoint will be marked")\n\n        ema_improved = ema_score < best_ema_score - 1e-6\n        raw_improved = raw_score < best_raw_score - 1e-6\n        history.append(dict(\n            epoch=epoch, train_loss=epoch_loss / train.G,\n            sam_perturbed_loss=epoch_perturbed_loss / train.G,\n            sam_first_grad_norm=epoch_grad_norm / max(1, epoch_updates),\n            optimizer_steps=step,\n            ema_val_loss=ema_clean, ema_val_swap_loss=ema_swap,\n            ema_selection_score=ema_score, ema_metrics=ema_metrics,\n            raw_val_loss=raw_clean, raw_val_swap_loss=raw_swap,\n            raw_selection_score=raw_score, raw_metrics=raw_metrics,\n            ema_improved=ema_improved, raw_improved=raw_improved,\n        ))\n\n        # Clone both states before saving.  The counterpart state is retained\n        # only as provenance; `state` is exactly the weight family being tested.\n        current_raw = {k: v.detach().cpu().clone() for k, v in model.state_dict().items()}\n        current_ema = {k: v.detach().cpu().clone() for k, v in ema_model.state_dict().items()}\n        common = dict(args=saved_args, R=train.R, gm=gm, gs=gs,\n                      method=f"v3_{variant}", struct_mode=args.struct_mode,\n                      training_complete=False, run_signature=run_signature,\n                      provenance=provenance, dual_selection=True,\n                      optimizer_name="sam_adamw", sam_rho=args.sam_rho, sam_replay_rng=True)\n        if ema_improved:\n            best_ema_score, best_ema_epoch = ema_score, epoch\n            atomic_save(dict(\n                state=current_ema, raw_state=current_raw, ema_state=current_ema,\n                thr=ema_threshold, best_epoch=epoch, weight_source="ema",\n                selection_model="ema", selection_score=ema_score, **common,\n            ), ema_checkpoint_path)\n        if raw_improved:\n            best_raw_score, best_raw_epoch = raw_score, epoch\n            atomic_save(dict(\n                state=current_raw, raw_state=current_raw, ema_state=current_ema,\n                thr=raw_threshold, best_epoch=epoch, weight_source="raw",\n                selection_model="raw", selection_score=raw_score, **common,\n            ), raw_checkpoint_path)\n        bad = 0 if (ema_improved or raw_improved) else bad + 1\n        write_json(record_path, dict(\n            complete=False, run_signature=run_signature, provenance=provenance,\n            best_ema_epoch=best_ema_epoch, best_raw_epoch=best_raw_epoch,\n            best_ema_score=best_ema_score, best_raw_score=best_raw_score,\n            history=history,\n        ))\n        print(f"Epoch {epoch:02d} | loss {epoch_loss/train.G:.4f} | "\n              f"EMA val {ema_clean:.4f} ({\'best\' if ema_improved else \'keep\'}) | "\n              f"Raw val {raw_clean:.4f} ({\'best\' if raw_improved else \'keep\'})", flush=True)\n        if bad >= args.patience:\n            break\n\n    for checkpoint_path in (ema_checkpoint_path, raw_checkpoint_path):\n        finished = torch.load(checkpoint_path, map_location="cpu", weights_only=True)\n        finished["training_complete"] = True\n        atomic_save(finished, checkpoint_path)\n    write_json(record_path, dict(\n        complete=True, run_signature=run_signature, provenance=provenance,\n        checkpoint_sha256={"ema": sha256(ema_checkpoint_path), "raw": sha256(raw_checkpoint_path)},\n        best_ema_epoch=best_ema_epoch, best_raw_epoch=best_raw_epoch,\n        best_ema_score=best_ema_score, best_raw_score=best_raw_score,\n        selection=args.select, history=history, train_seconds=time.time() - start,\n    ))\n    print(f"Saved EMA: {ema_checkpoint_path}", flush=True)\n    print(f"Saved Raw: {raw_checkpoint_path}", flush=True)\n    del model, teacher, optimizer, schedule\n    gc.collect()\n    if device.type == "cuda":\n        torch.cuda.empty_cache()\n\n\ndef main(argv=None):\n    args = build_parser().parse_args(argv)\n    validate_args(args)\n    args.determinism = configure_determinism(args.deterministic)\n    import torch\n\n    if args.device == "cuda" and not torch.cuda.is_available():\n        raise RuntimeError("Enable a GPU, or explicitly use --device cpu")\n    args.graph_dir, args.output_dir = args.graph_dir.resolve(), args.output_dir.resolve()\n    args.output_dir.mkdir(parents=True, exist_ok=True)\n    domain = next(item for item in TRAINING_DOMAINS.values()\n                  if (item["train_file"], item["val_file"]) == (args.train_file, args.val_file))\n    sources = [REPO / domain["train_source"], REPO / domain["val_source"]]\n    graphs = [args.graph_dir / f"{name}.pkl" for name in (args.train_file, args.val_file)]\n    data_hashes = {str(p): sha256(p) for p in sources + graphs}\n    pending = []\n    for variant in args.variants:\n        for seed in args.seeds:\n            path = args.output_dir / f"v3_{variant}_ema_seed{seed}.pt"\n            record_path = args.output_dir / f"v3_{variant}_dual_seed{seed}.train.json"\n            sig, provenance = signature(args, variant, seed, data_hashes)\n            if args.skip_completed and completed_run(path, record_path, sig):\n                print(f"Skip completed: {path.name}", flush=True)\n            elif (path.exists() or record_path.exists()) and not args.overwrite:\n                raise FileExistsError(f"Incomplete/different run: {path}. Use a new output directory "\n                                      "or --overwrite; a saved best epoch is not proof training finished.")\n            else:\n                pending.append((variant, seed, sig, provenance))\n    if not pending:\n        return\n    train = load_graph(graphs[0], sources[0], args.num_rel, args.struct_mode)\n    val = load_graph(graphs[1], sources[1], train.R, args.struct_mode)\n    for variant, seed, sig, provenance in pending:\n        train_one(args, variant, seed, train, val, sig, provenance)\n\n\nif __name__ == "__main__":\n    main()\n',encoding='utf-8')
compile(DUAL_TRAINER.read_text(encoding='utf-8'),str(DUAL_TRAINER),'exec')
print('Wrote:',DUAL_TRAINER)
```
